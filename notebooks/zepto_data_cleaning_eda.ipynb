```python
df["discount_amount"] = df["mrp"] - df["discounted_selling_price"]
df.head()
```

```python
# Keep the raw data unchanged by creating a separate cleaned copy
df_clean = df.copy()

df_clean["mrp"] = df_clean["mrp"] / 100
df_clean["discounted_selling_price"] = (
    df_clean["discounted_selling_price"] / 100
)
```

```python
#investigating the full dataset to see how many rows are matching and how many mismatching.
# Count matching and mismatching percentages
match_count = (
    df_clean["discount_percent"]
    == df_clean["calculated_discount_percent"]
).sum()

mismatch_count = (
    df_clean["discount_percent"]
    != df_clean["calculated_discount_percent"]
).sum()

print("Matching rows:", match_count)
print("Mismatching rows:", mismatch_count)
```

```python
import numpy as np

df_clean["is_truncated"] = (
    df_clean["discount_percent"]
    == np.floor(df_clean["calculated_discount_percent"])
)

print(df_clean["is_truncated"].value_counts())#This pattern strongly suggests that the original percentage was truncated to a whole number, rather than rounded to the nearest integer.
```

```python
df_clean.loc[
    ~df_clean["is_truncated"],#~ operator reverses the Boolean condition, so it selects rows where is_truncated is False
    [
        "name",
        "mrp",
        "discounted_selling_price",
        "discount_percent",
        "calculated_discount_percent",
        "discount_difference"
    ]
]
```

```python
'''One product (Cherry Blossom Liquid Shoe Polish) had MRP and selling price of 0, which is impossible for an in-stock item 
and likely a data entry error. With no way to recover the true price, and only 1 of 3,732 rows affected, I dropped it.'''
df = df.dropna(subset=["mrp", "discounted_selling_price"])
df = df.reset_index(drop=True)
df.shape
```

```python
# Final cleaning on df_clean 
# 1 - Drop the row with impossible price
df_clean = df_clean.dropna(subset=["mrp", "discounted_selling_price"])

# 2- Drop exact duplicate rows
df_clean = df_clean.drop_duplicates()

# 3 -Recalculate discount_amount in rupees (mrp and price are already in rupees)
df_clean["discount_amount"] = (df_clean["mrp"] - df_clean["discounted_selling_price"]).round(2)

# 4 - SQL-friendly columns
df_clean = df_clean.drop(columns=["is_truncated"])
df_clean["out_of_stock"] = df_clean["out_of_stock"].astype(int)   # 1 = out of stock
df_clean = df_clean.reset_index(drop=True)
df_clean.insert(0, "product_id", range(1, len(df_clean) + 1))

# 5 - Checks
print(df_clean.shape)                      # expected: (3729, 14)
print(df_clean.isnull().sum())             # only weight_in_gms should show 4
print(df_clean[["mrp", "discount_amount"]].head(2))   # Onion: 25.0 and 4.0

# 6 - Save the cleaned file
df_clean.to_csv("../data/zepto_cleaned.csv", index=False, encoding="utf-8")
```

```python

import pandas as pd
import mysql.connector

# 1. Read CSV
df = pd.read_csv("../data/zepto_cleaned.csv", encoding="utf-8")
df = df.where(pd.notna(df), None)

print("CSV rows:", len(df))

# 2. Connect to MySQL
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password=input("MySQL password: "),
    database="zepto_analytics"
)

cursor = conn.cursor()

columns = [
    "product_id", "category", "name", "mrp",
    "discount_percent", "available_quantity",
    "discounted_selling_price", "weight_in_gms",
    "out_of_stock", "quantity", "discount_amount",
    "calculated_discount_percent", "stock_status",
    "discount_difference"
]

# Convert pandas/numpy values into MySQL-friendly values
rows = [
    tuple(None if pd.isna(v) else v.item() if hasattr(v, "item") else v
          for v in row)
    for row in df[columns].itertuples(index=False, name=None)
]

placeholders = ", ".join(["%s"] * len(columns))
column_names = ", ".join(f"`{c}`" for c in columns)

sql = f"""
INSERT INTO zepto_cleaned ({column_names})
VALUES ({placeholders})
"""

try:
    # Clear old rows, then load all CSV rows
    cursor.execute("DELETE FROM zepto_cleaned")
    cursor.executemany(sql, rows)
    conn.commit()

    cursor.execute("SELECT COUNT(*) FROM zepto_cleaned")
    print("Rows in MySQL after import:", cursor.fetchone()[0])
    print("Rows inserted:", cursor.rowcount)

except Exception as e:
    conn.rollback()
    print("Import failed. Changes rolled back.")
    print("Error:", e)

finally:
    cursor.close()
    conn.close()
```
